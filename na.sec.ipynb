```python
import torch
x=torch.tensor(2.0,requires_grad=True)
y=x**2+2*x
y.backward()
x.grad
```

```python
import torch
x=torch.tensor(2.0,requires_grad=True)
y=torch.tensor(4.0)
z=x**2+y**2+2*x*y
z.backward()
print(x.grad)
print(y.grad)
```

```python
import torch
x=torch.tensor(2.0,requires_grad=True)
y=torch.tensor(4.0,requires_grad=True)
z=x**2+y**2+2*x*y
z.backward()
print(x.grad)
print(y.grad)
```

```python
x=torch.tensor(2.0)
y_true=torch.tensor(1.0)
w=torch.tensor(.1,requires_grad=True)

loss=(x*w - y_true)**2
loss.backward()

print(x.grad)
print(w.grad)
```

```python
x=torch.tensor(2.0)
y_true=torch.tensor(1.0)
w=torch.tensor(.1,requires_grad=True)
learning_rate=.1
loss=(x*w - y_true)**2
loss.backward()
print(w.grad)

with torch.no_grad():
     w-=learning_rate*w.grad
     w.grad.zero_()

print(w)
print(w.grad)
```
